## 1. Question

*The research question and the decision it supports.*

## Research Question

Can observable search performance, engagement, and freshness signals be used to rank website pages by priority for content refresh?

### Decision Supported

The goal is to help an SEO or content team decide which existing website pages should be reviewed first for a potential content refresh. The output will be a ranked list of pages with a refresh-priority score.

### Unit of Analysis

One website page per row.

### Intended Use

The ranking is a decision-support tool for prioritizing manual review, not a guarantee that refreshing a page will improve its search performance.

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

## Methodology

### Assumptions

This study assumes that observable search performance, engagement, content freshness, and page characteristics can help identify pages that deserve earlier review for a potential content refresh. The resulting score is a prioritization aid rather than a prediction of guaranteed improvement.

### Features

Candidate features include search impressions, clicks, click-through rate (CTR), average search position, freshness-related signals, and content-depth indicators where available. Features will be selected based on their availability and documented meaning in the chosen dataset.

### Label Definition

The target is `is_declining_label`, derived from `trend_direction`, where a value of `down` indicates a declining page. This is a proxy label describing the observed trend in the available data; it does not represent a verified future outcome.

### Baseline

The rule-based baseline combines visibility, freshness risk, position opportunity, and content-depth gap:

`baseline_refresh_score = 0.40 × visibility_score + 0.30 × freshness_risk_score + 0.25 × position_opportunity_score + 0.05 × depth_gap_score`

The baseline and machine-learning model should be evaluated on the same held-out data.

### Model and Validation Design

The model will be compared with the rule-based baseline using the same validation split. Where available, a client-holdout split will be used so that pages from held-out clients do not appear in the training data. This helps assess generalization to unseen clients.

### Leakage Checks

The features `trend_direction` and `trend_pct` must not be used as predictors because they are directly associated with the target label. Other features will be checked for target leakage and for availability at the time a refresh decision would be made.

### Evaluation Metrics

ROC AUC and average precision (AP) will assess ranking and classification quality. Precision@50 will measure the proportion of declining pages among the 50 highest-priority pages. These metrics will be reported for the baseline and model using the same split.

## 5. Limitations

*What this work cannot claim.*

## Limitations

1. **Proxy target:** The label identifies an observed declining trend; it is not a verified future outcome or a direct measure of refresh success.

2. **No causal evidence:** The analysis is observational. It cannot prove that refreshing a page will improve clicks, impressions, traffic, or rankings.

3. **Dataset scope:** Results apply to the dataset, features, and time windows actually used. They should not be generalized to the entire FlyRank warehouse unless that broader analysis was performed.

4. **Potential distribution shift:** Performance may change across clients, industries, search markets, and time periods.

5. **Feature availability:** Some signals may be incomplete, noisy, or unavailable when a real refresh decision is made.

6. **Ranking is not a final decision:** High-priority pages require human review of content quality, search intent, business relevance, and whether a refresh is appropriate.

7. **Privacy:** Published artifacts must not reveal client identities, private URLs, private queries, credentials, or other identifying information.

Future work should test the ranking prospectively and evaluate whether acting on the recommendations leads to measurable improvements.

## 6. Ranked recommendations

## Ranked Recommendations: Content Refresh Playbook

The model's output should be used to identify pages for review, not to automatically schedule content changes. The following playbook translates the ranking into actionable steps.

### Priority 1 — Review the highest-ranked pages

Start with pages receiving the highest refresh-priority scores. Verify that the declining-performance label and supporting signals are correct before deciding whether a refresh is appropriate.

**Action:** Review search impressions, clicks, click-through rate, average position, and the available freshness indicators.

### Priority 2 — Diagnose the underlying opportunity

Determine why each selected page may need attention. Possible explanations include outdated information, a mismatch with search intent, weak search-result presentation, or incomplete coverage of the topic.

**Action:** Have a content editor inspect the page and document the reason for recommending a refresh.

### Priority 3 — Match the intervention to the diagnosis

Do not apply the same intervention to every page. Update factual information where necessary, improve relevance to search intent, or revise the title and description when the evidence indicates a search-result presentation problem.

**Action:** Record the proposed change and the expected outcome before publishing.

### Priority 4 — Track outcomes after the refresh

Establish a pre-refresh performance baseline and monitor the same metrics after publication. Account for seasonality, broader search changes, and other interventions where possible.

**Action:** Compare post-refresh performance with the pre-refresh period and, where feasible, an appropriate comparison group.

### Priority 5 — Reassess the ranking

Use observed outcomes to evaluate whether the prioritization system is useful. Revisit feature choices and scoring weights as additional evidence becomes available.

**Action:** Periodically review model performance and adjust the workflow when the ranking does not produce useful editorial decisions.

### Decision Rule

Use the score to decide **which pages to inspect first**. Make the final refresh decision only after human review. The ranking alone does not establish that a page refresh will improve search performance.
