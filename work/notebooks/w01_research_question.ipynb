## 2. The question: decision, action, cost of a wrong call

### Decision, Action, and Cost of a Wrong Recommendation:

### Decision

The goal is to help determine **which website pages should be refreshed first**. Since content teams have limited time and resources, they cannot update every page. This project aims to prioritize pages that are most likely to benefit from a content refresh.

### Who Acts on It?

The recommendations will be used by **SEO specialists** and **content editors**. They can use the ranked list to decide which pages to update, optimize, or monitor first.

### Action:

For each page, the model produces a **Refresh Priority Score**. Pages with the highest scores are reviewed and refreshed before lower-priority pages, allowing the team to focus on content with the greatest potential impact.

### Cost of a Wrong Recommendation:

**False Positive (High priority but doesn't need refreshing):**
- Wastes editorial time and effort.
- Uses resources on pages with little expected benefit.

**False Negative (Needs refreshing but not recommended):**
- Valuable pages may continue to lose search visibility.
- Organic traffic and user engagement may decline.
- Potential business opportunities may be missed.

### Dataset Summary

The starter dataset contains **30,000 website pages** described by **44 features**, including search performance, engagement metrics, content metadata, and freshness signals.

A few observations that support the **Refresh / Content Opportunity Scoring** lane:

- The average page receives **5,200 impressions** over 90 days, but the median is only **731**, indicating that page visibility varies widely across the dataset.
- The median **CTR is 0.07%**, while the average is **0.51%**, suggesting many pages have low click-through rates and may benefit from optimization.
- The median page age is **236 days**, with pages ranging from **90 to 564 days** old, indicating that some content is significantly older than others and may require refreshing.

These observations suggest that there is substantial variation in visibility, engagement, and content freshness, making it worthwhile to build a scoring system that prioritizes pages for content refresh.

## What My Work Can and Cannot Claim

### What this project can claim

- It identifies **observed patterns** in search performance, engagement, and content freshness.

### What this project cannot claim

- It does **not** prove that refreshing a page will cause better search rankings or increased traffic.

- It does **not** guarantee that every recommended page will improve after being refreshed.

The recommendations produced by this project should be used as **decision support** rather than definitive predictions.
